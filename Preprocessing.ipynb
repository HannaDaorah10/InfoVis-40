### Variable descriptions

In terms of variable type and measurement scale, the variables in this dataset can be described as follows:

- **Categorical / Nominal variables**:  
  `PRODUCT`  
  Indicates the type of energy source (e.g., “Wind”, “Oil”, “Solar”). Useful for analyzing the energy mix and comparing the growth of renewables versus fossil fuels.

- **Continuous / Ratio variables**:  
  `VALUE`, `GDP per capita`, `GDP total`  
  These are quantitative variables that measure production quantity (in GWh) and economic indicators.  
  - `VALUE` shows the monthly energy production per source.  
  - `GDP per capita` and `GDP total` help examine the relationship between a country's economic status and its investment in renewable energy.

- **Interval variable**:  
  `YEAR`  
  Represents the year of energy production. Useful for identifying long-term trends in energy production and renewable adoption.

- **Percentage-based Continuous variables (Ratio)**:  
  `Access to electricity (% of population)`,  
  `Access to clean fuels for cooking (% of population)`,  
  `Renewable energy share in total final energy consumption (%)`  
  These percentage variables indicate broader access and sustainability metrics:  
  - `Access to electricity` reflects how widespread electricity is within a country.  
  - `Access to clean fuels for cooking` captures sustainable energy use beyond electricity.  
  - `Renewable energy share` illustrates how much of a country’s energy consumption comes from renewable sources.
